# Gravitational Wave Capture

This module presents the pbh capture via gravitational wave emission simulations used in the paper.

```python
%load_ext autoreload
%autoreload 2
import src.GWCapture.simulation as gwcsim
import numpy as np
import astropy.constants as const
import astropy.units as u
```

We simulate a typical neutron star with following parameters. One can run this cell to recreate the paper results or change it to simulate another configuration.

```python

star = 'Typical_Neutron_Star'
M_star = 1.4 * const.M_sun.value # in M_sun
R_star = 11e3 # in meters

gridsize = 100
m_pbh_values = np.logspace(-10, 2, gridsize) * const.M_sun.value # in M_sun
v_inf_values = np.logspace(3, 6, gridsize) # in m/s
r = 8 # kpc

conf_dict = {
    'star': star,
    'M_star': M_star,
    'R_star': R_star,
    'm_pbh_values': m_pbh_values,
    'v_inf_values': v_inf_values,
    'gridsize': gridsize,
    'r': r
}
```

```python
capture = gwcsim.GWCapture(conf_dict)
```
